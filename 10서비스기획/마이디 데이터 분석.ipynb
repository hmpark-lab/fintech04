```python
import pandas as pd
import glob
import os

folder_path = r"C:\Users\hmabc\Downloads\myd_reviews"

file_list = sorted(glob.glob(os.path.join(folder_path, "*.csv")))

print("찾은 파일 수:", len(file_list))
print(file_list[:3])

df_list = []
failed_files = []

for file in file_list:
    try:
        temp = pd.read_csv(file, encoding="utf-16")
        month = os.path.basename(file).split("_")[-1].replace(".csv", "")
        temp["review_month"] = month
        temp["source_file"] = os.path.basename(file)
        df_list.append(temp)
        print("성공:", os.path.basename(file), temp.shape)

    except Exception as e:
        failed_files.append((os.path.basename(file), str(e)))
        print("실패:", os.path.basename(file), e)

print("성공 파일 수:", len(df_list))
print("실패 파일 수:", len(failed_files))

if len(df_list) > 0:
    df_reviews = pd.concat(df_list, ignore_index=True)
    print("최종 데이터 크기:", df_reviews.shape)
    display(df_reviews.head())
else:
    print("합칠 데이터가 없습니다. file_list 경로를 다시 확인하세요.")
```

```python
keep_cols = [
    'review_month',
    'Review Submit Date and Time',
    'Star Rating',
    'Review Text',
    'Reviewer Language',
    'Device',
    'App Version Name'
]

df_reviews = df_reviews[keep_cols].copy()
```

```python
df_reviews['Review Text'] = df_reviews['Review Text'].fillna('')

df_reviews = df_reviews[
    df_reviews['Review Text'].str.strip() != ''
]
```

```python
df_reviews['Review Submit Date and Time'] = pd.to_datetime(
    df_reviews['Review Submit Date and Time']
)
```
